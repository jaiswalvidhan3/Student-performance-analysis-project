# Student Performance Analytics Engine

Analysis of a 500 x 6 marks matrix: subject and student statistics, rankings and outlier detection.

## 1. Imports, seed and marks matrix

```python
import numpy as np

np.random.seed(42)

subjects = ["Maths", "Physics", "Chemistry", "English", "CS", "Biology"]

# 500 students, 6 subjects, marks out of 100
marks = np.random.normal(65, 15, (500, 6))
marks = marks.round().clip(0, 100)

print(marks.shape)
print(marks[:5])
```

The seed keeps the random marks the same every time the notebook is run.

## 2. Sum, mean, median, std and var

axis=0 gives one value per subject (down the columns). axis=1 gives one value per student (across the row).

```python
print("Per subject (axis=0)")
print(subjects)
print("sum   :", marks.sum(axis=0))
print("mean  :", marks.mean(axis=0).round(2))
print("median:", np.median(marks, axis=0))
print("std   :", marks.std(axis=0).round(2))
print("var   :", marks.var(axis=0).round(2))
```

```python
print("Per student (axis=1), first 5 students")
print("sum   :", marks.sum(axis=1)[:5])
print("mean  :", marks.mean(axis=1)[:5].round(2))
print("median:", np.median(marks, axis=1)[:5])
print("std   :", marks.std(axis=1)[:5].round(2))
print("var   :", marks.var(axis=1)[:5].round(2))
```

## 3. Z-scores, outliers and ranking

z = (mark - subject mean) / subject std. The mean and std have shape (6,), so NumPy broadcasts them across all 500 rows.

```python
mean = marks.mean(axis=0)
std = marks.std(axis=0)

z = (marks - mean) / std
print(z.shape)

# outlier if the z-score is above 2 or below -2
outliers = np.abs(z) > 2
print("Total outlier marks:", outliers.sum())
print("Outliers per subject:", outliers.sum(axis=0))

rows, cols = np.where(outliers)
print("First 5 outliers (student, subject, mark):")
for i in range(5):
    print(rows[i], subjects[cols[i]], marks[rows[i], cols[i]])
```

```python
total = marks.sum(axis=1)

# rank 1 = highest total
order = np.argsort(-total)
rank = np.argsort(order) + 1

print("Top 5 students (index, total):")
for i in order[:5]:
    print(i, total[i])
```

## 4. Add totals with hstack and export CSV

```python
from google.colab import files

total = total.reshape(500, 1)
rank = rank.reshape(500, 1)

final = np.hstack((marks, total, rank))
print(final.shape)
print(final[:3])

header = ",".join(subjects) + ",Total,Rank"
np.savetxt("student_marks.csv", final, delimiter=",", header=header, comments="", fmt="%d")

files.download("student_marks.csv")
```
